#**SORTING**

Sorting adalah proses mengurutkan data berdasarkan aturan tertentu.

Pada tugas ini, data diurutkan secara ascending,
yaitu dari nilai terkecil ke nilai terbesar.

Tujuan percobaan ini adalah untuk mengetahui
waktu yang diperlukan beberapa algoritma sorting
dalam mengurutkan data.

    Jumlah data yang digunakan :

    100, 200, 300, 400, 500

Data dibuat secara acak.

Rumus pengukuran waktu :

waktu = selesai - mulai

```python
import random
import time

ukuran_data = [100, 200, 300, 400, 500]
```

#**1. BUBBLE SORT**

Bubble sort membandingkan dua data yang berdekatan.

Jika urutannya salah, kedua data akan ditukar.
Proses dilakukan berulang sampai seluruh data terurut.

```python
def bubble_sort(arr):
    a = arr.copy()

    for i in range(len(a)):
        for j in range(0, len(a)-i-1):
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]

    return a
```

```python
#Pengujian waktu Bubble Sort
for n in ukuran_data:
    data = random.sample(range(1, 10000), n)

    mulai = time.time()
    bubble_sort(data)
    selesai = time.time()

    print("Jumlah data =", n, "waktu =", selesai - mulai)
```

#**2. SELECTION SORT**

Selection sort mencari nilai terkecil
dan menempatkannya pada posisi yang benar.

```python
def selection_sort(arr):
    a = arr.copy()

    for i in range(len(a)):
        min_idx = i

        for j in range(i+1, len(a)):
            if a[j] < a[min_idx]:
                min_idx = j

        a[i], a[min_idx] = a[min_idx], a[i]

    return a
```

```python
#Pengujian waktu Selection Sort
for n in ukuran_data:
    data = random.sample(range(1, 10000), n)

    mulai = time.time()
    selection_sort(data)
    selesai = time.time()

    print("Jumlah data =", n, "waktu =", selesai - mulai)
```

#**3. INSERTION SORT**

Insertion sort menganggap bagian kiri data
sudah terurut.

Kemudian satu data berikutnya disisipkan
ke posisi yang benar.

```python
def insertion_sort(arr):
    a = arr.copy()

    for i in range(1, len(a)):
        nilai_awal = a[i]
        hole = i

        while hole > 0 and a[hole-1] > nilai_awal:
            a[hole] = a[hole-1]
            hole = hole - 1

        a[hole] = nilai_awal

    return a
```

```python
#Pengujian waktu Insertion Sort
for n in ukuran_data:
    data = random.sample(range(1, 10000), n)

    mulai = time.time()
    insertion_sort(data)
    selesai = time.time()

    print("Jumlah data =", n, "waktu =", selesai - mulai)
```

#**4. MERGE SORT**

Merge sort membagi data menjadi dua bagian.

Setelah itu setiap bagian diurutkan,
kemudian digabungkan kembali.

```python
def merge_sort(a):
    if len(a) < 2:
        return a

    mid = len(a) // 2
    left = merge_sort(a[:mid])
    right = merge_sort(a[mid:])

    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result
```

```python
#Pengujian waktu Merge Sort
for n in ukuran_data:
    data = random.sample(range(1, 10000), n)

    mulai = time.time()
    merge_sort(data)
    selesai = time.time()

    print("Jumlah data =", n, "waktu =", selesai - mulai)
```

#**5. QUICK SORT**

Quick sort memilih satu data sebagai pivot.

Data yang lebih kecil diletakkan di kiri,
dan data yang lebih besar diletakkan di kanan.

```python
def quick_sort(arr):
    a = arr.copy()

    def partition(A, start, end):
        pivot = A[end]
        pIndex = start

        for i in range(start, end):
            if A[i] <= pivot:
                A[i], A[pIndex] = A[pIndex], A[i]
                pIndex += 1

        A[pIndex], A[end] = A[end], A[pIndex]
        return pIndex

    def quick(A, start, end):
        if start < end:
            pIndex = partition(A, start, end)
            quick(A, start, pIndex - 1)
            quick(A, pIndex + 1, end)

    quick(a, 0, len(a) - 1)
    return a
```

```python
#Pengujian waktu Quick Sort
for n in ukuran_data:
    data = random.sample(range(1, 10000), n)

    mulai = time.time()
    quick_sort(data)
    selesai = time.time()

    print("Jumlah data =", n, "waktu =", selesai - mulai)
```

**Kesimpulan**

Semakin besar jumlah data,
waktu pengurutan akan semakin bertambah.

Bubble sort, selection sort, dan insertion sort
cenderung membutuhkan waktu lebih lama
dibanding merge sort dan quick sort.
