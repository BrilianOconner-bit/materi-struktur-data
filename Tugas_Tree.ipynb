#**1.Menampilkan tree**

Tree adalah struktur data hirarkis yang terdiri dari node (simpul) yang saling terhubung.

Struktur umum:

Root → node paling atas
Parent → node induk
Child → node anak
Leaf → node tanpa anak


```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def preorder(node):
    if node:
        print(node.data, end=" ")
        preorder(node.left)
        preorder(node.right)

# Membuat tree
root = Node("A")
root.left = Node("B")
root.right = Node("C")
root.left.left = Node("D")
root.left.right = Node("E")

print("Preorder Traversal:")
preorder(root)
```

#**2.Program Trace Parse Tree**
Persamaan Matematika harus dipecah menjadi list token agar dapat dibuat menjadi parse tree.

Token-token untuk persamaan Matematika tersebut antara lain:

- Kurung buka, kurung buka ini menandakan ekspressi baru yang akan dioperasi, oleh karena
itu perlu dibuat subtree baru untuk menyelesaikan ekspressi baru ini
kurung tutup,
- kurung tutup ini menandakan akhir dari ekspressi matematika

- operand, operand ini yang akan jadi leaf node dan children dari operator

- operator, operator merupakan parent dan punya left maupun right child

```python
def stack():
    return []

def push(s, data):
    s.append(data)

def pop(s):
    return s.pop()
```

```python
class BinaryTree:

    def __init__(self, root):
        self.key = root
        self.leftChild = None
        self.rightChild = None

    def insertLeft(self, new_node):
        if self.leftChild is None:
            self.leftChild = BinaryTree(new_node)
        else:
            t = BinaryTree(new_node)
            t.leftChild = self.leftChild
            self.leftChild = t

    def insertRight(self, new_node):
        if self.rightChild is None:
            self.rightChild = BinaryTree(new_node)
        else:
            t = BinaryTree(new_node)
            t.rightChild = self.rightChild
            self.rightChild = t

    def getLeftChild(self):
        return self.leftChild

    def getRightChild(self):
        return self.rightChild

    def setRootVal(self, obj):
        self.key = obj

    def getRootVal(self):
        return self.key
```

```python
def buildParseTree(mathExp):
    tokenList = mathExp.split()
    parentStack = stack()
    expTree = BinaryTree(' ')

    push(parentStack, expTree)
    print("Token List:", tokenList)

    currentTree = expTree

    for i in tokenList:

        if i == '(':
            print('if 1 ( buka ):', i)
            currentTree.insertLeft(' ')
            push(parentStack, currentTree)
            currentTree = currentTree.getLeftChild()

        elif i not in ['+', '-', '*', '/', ')']:
            print('if 2 ( operand ):', i)
            currentTree.setRootVal(int(i))
            parent = pop(parentStack)
            currentTree = parent

        elif i in ['+', '-', '*', '/']:
            print('if 3 ( operator ):', i)
            currentTree.setRootVal(i)
            currentTree.insertRight(' ')
            push(parentStack, currentTree)
            currentTree = currentTree.getRightChild()

        elif i == ')':
            print('if 4 ( tutup ):', i)
            currentTree = pop(parentStack)

        else:
            raise ValueError("Token tidak valid")

    return expTree
```

```python
pt = buildParseTree(' ( 3 + ( 4 * 5 ) ) ')
```

```python
import operator

def evaluate(parse_tree):
    opers = {
        '+': operator.add,
        '-': operator.sub,
        '*': operator.mul,
        '/': operator.truediv
    }

    left = parse_tree.getLeftChild()
    right = parse_tree.getRightChild()

    if left and right:
        fn = opers[parse_tree.getRootVal()]
        return fn(evaluate(left), evaluate(right))
    else:
        return parse_tree.getRootVal()
```

```python
evaluate(pt)
```
