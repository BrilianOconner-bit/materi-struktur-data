#**HASHING**


Hashing adalah teknik untuk menyimpan data ke dalam
sebuah tabel dengan menggunakan fungsi hash.
Fungsi hash mengubah sebuah data (key) menjadi
sebuah index atau alamat pada tabel.

Tujuan hashing adalah agar proses pencarian,
penambahan, dan penghapusan data dapat dilakukan
lebih cepat.

Dalam hashing, terkadang dua nilai berbeda dapat
menghasilkan posisi yang sama. Kondisi tersebut
disebut collision.Untuk mengatasi collision dapat digunakan beberapa metode,
di antaranya chaining, linear probing, dan quadratic probing.

Pada tugas ini digunakan ukuran tabel hash sebanyak 11
dengan fungsi hash :

*h(x) = x mod 11*

Data yang digunakan :

54, 26, 93, 17, 77, 31, 44, 55, 20

# **1. CLOSE ADDRESS (CHAINING)**

Pada metode chaining, jika beberapa data mempunyai
index yang sama maka data tersebut disimpan
pada index yang sama dalam bentuk list.

Langkah kerja :

1. Hitung nilai hash
2. Jika index kosong, data langsung disimpan
3. Jika terjadi collision, data dimasukkan
   ke list pada index tersebut

```python
table = [[] for _ in range(11)]

def hash_func(x):
    return x % 11

def insert(table, key):
    index = hash_func(key)
    table[index].append(key)

data = [54, 26, 93, 17, 77, 31, 44, 55, 20]

for x in data:
    insert(table, x)

print("Hash table chaining =", table)
```

CRUD pada chaining
*   Create  : menambah data
*   Read    : mencari data
* Update  : mencari data lama lalu menggantinya
* Delete  : mencari data lalu menghapusnya



```python
def search(table, key):
    index = hash_func(key)

    if key in table[index]:
        return f"{key} ditemukan pada index {index}"

    return "data tidak ditemukan"

def update(table, old_key, new_key):
    index = hash_func(old_key)

    if old_key in table[index]:
        table[index].remove(old_key)
        table[hash_func(new_key)].append(new_key)
        return "data berhasil diupdate"

    return "data tidak ditemukan"

def delete(table, key):
    index = hash_func(key)

    if key in table[index]:
        table[index].remove(key)
        return "data berhasil dihapus"

    return "data tidak ditemukan"

print(search(table, 44))
print(update(table, 55, 25))
print(table)

print(delete(table, 20))
print(table)
```

# **2. OPEN ADDRESS**

Pada metode open address, data disimpan langsung
ke dalam tabel hash.

Jika terjadi collision, pencarian slot kosong dilakukan
dengan aturan tertentu.

Pada bagian ini digunakan dua fungsi hash, yaitu :

1. Fungsi hash biasa
   h(x) = x mod 11

2. Mid Square Method

#**2.1 LINEAR PROBING**


**Hash Biasa**

Pada linear probing, jika terjadi collision maka
pencarian slot kosong dilakukan secara berurutan.

```python
table = [None] * 11

def hash_func(x):
    return x % 11

def insert(table, key):
    index = hash_func(key)

    while table[index] is not None:
        index = (index + 1) % len(table)

    table[index] = key

data = [54, 26, 93, 17, 77, 31, 44, 55, 20]

for x in data:
    insert(table, x)

print("Linear probing =", table)
```

**CRUD pada linear probing**

Pencarian dimulai dari index hasil hash.
Jika tidak cocok maka dicek ke index berikutnya.

```python
def search(table, key):
    index = hash_func(key)
    start = index

    while table[index] is not None:
        if table[index] == key:
            return f"{key} ditemukan pada index {index}"

        index = (index + 1) % len(table)

        if index == start:
            break

    return "data tidak ditemukan"

def delete(table, key):
    index = hash_func(key)
    start = index

    while table[index] is not None:
        if table[index] == key:
            table[index] = None
            return "data berhasil dihapus"

        index = (index + 1) % len(table)

        if index == start:
            break

    return "data tidak ditemukan"

def update(table, old_key, new_key):
    delete(table, old_key)
    insert(table, new_key)

print(search(table, 44))

update(table, 55, 25)
print(table)

print(delete(table, 20))
print(table)
```

**Linear Probing dengan mid square method**

Pada bagian ini, fungsi hash yang digunakan adalah
mid square method, sedangkan proses pencarian
slot kosong tetap menggunakan linear probing.

```python
table = [None] * 11

def mid_square_hash(key, size):
    square = key * key
    s = str(square)

    mid = len(s) // 2

    if len(s) % 2 == 0:
        middle = s[mid - 1:mid + 1]
    else:
        middle = s[mid]

    return int(middle) % size

def insert_mid_linear(table, key):
    index = mid_square_hash(key, len(table))

    while table[index] is not None:
        index = (index + 1) % len(table)

    table[index] = key

for x in data:
    insert_mid_linear(table, x)

print("Linear probing mid square =", table)
```

#**2.2  QUADRATIC PROBING**




**Hash Biasa**

Pada quadratic probing, jika terjadi collision maka
pencarian slot kosong menggunakan pola kuadrat.

```python
table = [None] * 11

def insert(table, key):
    index = hash_func(key)

    if table[index] is None:
        table[index] = key
    else:
        i = 1

        while True:
            new_index = (index + i * i) % len(table)

            if table[new_index] is None:
                table[new_index] = key
                break

            i += 1

for x in data:
    insert(table, x)

print("Quadratic probing =", table)
```

**CRUD pada quadratic probing**

Proses pencarian dilakukan dengan pola index kuadrat
yang sama seperti proses insert.

```python
def search(table, key):
    index = hash_func(key)

    if table[index] == key:
        return f"{key} ditemukan pada index {index}"

    i = 1

    while i < len(table):
        new_index = (index + i * i) % len(table)

        if table[new_index] == key:
            return f"{key} ditemukan pada index {new_index}"

        i += 1

    return "data tidak ditemukan"

def delete(table, key):
    index = hash_func(key)

    if table[index] == key:
        table[index] = None
        return "data berhasil dihapus"

    i = 1

    while i < len(table):
        new_index = (index + i * i) % len(table)

        if table[new_index] == key:
            table[new_index] = None
            return "data berhasil dihapus"

        i += 1

    return "data tidak ditemukan"

def update(table, old_key, new_key):
    delete(table, old_key)
    insert(table, new_key)

print(search(table, 44))

update(table, 55, 25)
print(table)

print(delete(table, 20))
print(table)
```

**Quadratic probing dengan mid square method**

Pada bagian ini, fungsi hash yang digunakan adalah
mid square method, sedangkan pencarian slot kosong
tetap mengikuti pola quadratic probing.

```python
table = [None] * 11

def insert_mid_quadratic(table, key):
    index = mid_square_hash(key, len(table))

    if table[index] is None:
        table[index] = key
    else:
        i = 1

        while True:
            new_index = (index + i * i) % len(table)

            if table[new_index] is None:
                table[new_index] = key
                break

            i += 1

for x in data:
    insert_mid_quadratic(table, x)

print("Quadratic probing mid square =", table)
```

**Load Factor**

Load factor menunjukkan perbandingan antara jumlah data
yang tersimpan dengan ukuran tabel hash.

alpha = n / m

```python
n = len(data)
m = 11

alpha = n / m

print("Jumlah data =", n)
print("Ukuran tabel =", m)
print("Load factor =", alpha)
```

Karena alpha < 1, maka proses insert masih dapat dilakukan.

**Contoh insert ketika tabel penuh**

Jika seluruh slot tabel telah terisi,
maka data baru tidak dapat dimasukkan.

```python
table = [None] * 11

data_penuh = [54, 26, 93, 17, 77, 31, 44, 55, 20, 25, 36]

def insert(table, key):
    index = hash_func(key)
    start = index

    while table[index] is not None:
        index = (index + 1) % len(table)

        if index == start:
            print("Tabel penuh, data", key, "tidak dapat dimasukkan")
            return

    table[index] = key

for x in data_penuh:
    insert(table, x)

print("Isi tabel =", table)

insert(table, 48)
```

# **3. MID SQUARE METHOD**

Mid square method adalah fungsi hash
dengan langkah berikut :

1. Bilangan dikuadratkan
2. Ambil digit tengah
3. Hasil digit tengah dimoduluskan
   dengan ukuran tabel

```python
def mid_square_hash(key, size):
    square = key * key
    s = str(square)

    mid = len(s) // 2

    if len(s) % 2 == 0:
        middle = s[mid - 1:mid + 1]
    else:
        middle = s[mid]

    return int(middle) % size

data = [54, 26, 93, 17]

for x in data:
    print(x, "->", mid_square_hash(x, 11))
```

Kesimpulan

1. Chaining menyimpan collision pada index yang sama
2. Linear probing mencari slot kosong secara berurutan
3. Quadratic probing mencari slot kosong dengan pola kuadrat
4. Mid square method memakai digit tengah hasil kuadrat
