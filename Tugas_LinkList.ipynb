#**A. UNORDERED LINKED LIST**

Aturan soal
Search Cari angka tertentu 18
1.   Jika 18 ditemukan → sisipkan setelah angka 18
2.   Jika 18 tidak ditemukan → sisipkan di akhir list








**1. Jika 18 ditemukan → sisipkan setelah angka 18**

```python
# class node
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# class linked list
class LinkedList:
    def __init__(self):
        self.head = None

    def tambah_awal(self, data):
        node_baru = Node(data)
        node_baru.next = self.head
        self.head = node_baru

    def tampilkan(self):
        temp = self.head
        while temp:
            print(temp.data, end=" --> ")
            temp = temp.next
        print("None")

    def insert_18_unordered(self):
        temp = self.head

        while temp:
            if temp.data == 18:
                node_baru = Node(18)
                node_baru.next = temp.next
                temp.next = node_baru
                return
            temp = temp.next

        # jika 18 tidak ditemukan, sisipkan di akhir
        node_baru = Node(18)
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = node_baru


# contoh 18 ditemukan
ll = LinkedList()
ll.tambah_awal(6)
ll.tambah_awal(18)
ll.tambah_awal(7)

print("Sebelum:")
ll.tampilkan()

ll.insert_18_unordered()

print("Sesudah:")
ll.tampilkan()
```

**2. Jika 18 tidak ditemukan → sisipkan di akhir list**

```python
# class node
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# class linked list
class LinkedList:
    def __init__(self):
        self.head = None

    def tambah_awal(self, data):
        node_baru = Node(data)
        node_baru.next = self.head
        self.head = node_baru

    def tampilkan(self):
        temp = self.head
        while temp:
            print(temp.data, end=" --> ")
            temp = temp.next
        print("None")

    def insert_18_jika_tidak_ada_di_akhir(self):
        temp = self.head

        # SEARCH 18
        while temp:
            if temp.data == 18:
                print("18 ditemukan, tidak masuk kasus 2")
                return
            if temp.next is None:
                break
            temp = temp.next

        # insert di akhir
        node_18 = Node(18)
        temp.next = node_18

ll = LinkedList()
ll.tambah_awal(6)
ll.tambah_awal(10)
ll.tambah_awal(7)

print("Sebelum:")
ll.tampilkan()

ll.insert_18_jika_tidak_ada_di_akhir()

print("Sesudah:")
ll.tampilkan()
```

# **B. OrderList**

iterasi:
- selama data < sekarang → lanjut
- jika data sekarang > yang dicari → BERHENTI

```python
# class node
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# ordered linkedlist
class OrderedLinkedList:
    def __init__(self):
        self.head = None

    def tambah(self, data):
        node = Node(data)

        if self.head is None or data < self.head.data:
            node.next = self.head
            self.head = node
            return

        temp = self.head
        while temp.next and temp.next.data < data:
            temp = temp.next

        node.next = temp.next
        temp.next = node

    def tampilkan(self):
        temp = self.head
        while temp:
            print(temp.data, end=" --> ")
            temp = temp.next
        print("None")

    def insert_18_ordered(self):
        temp = self.head
        prev = None

        # ITERASI
        while temp and temp.data < 18:
            prev = temp
            temp = temp.next

        # jika ada 18
        if temp and temp.data == 18:
            node_18 = Node(18)
            node_18.next = temp.next
            temp.next = node_18

        # jika tidak ada 18
        else:
            node_18 = Node(18)
            node_18.next = temp
            if prev:
                prev.next = node_18
            else:
                self.head = node_18
```

```python
oll = OrderedLinkedList()
oll.tambah(7)
oll.tambah(10)
oll.tambah(18)
oll.tambah(25)

oll.tampilkan()
oll.insert_18_ordered()
oll.tampilkan()
```
